Now, let's convert some text into tokens using `tiktoken`.

```python
import tiktoken

# Sample text to tokenize
text = "Hello, how are you today? This is an example of tokenization using tiktoken."

# Load an encoding. For example, 'cl100k_base' is used by gpt-4, gpt-3.5-turbo, text-embedding-ada-002.
# You can choose other encodings based on the model you are working with.
encoding = tiktoken.get_encoding("cl100k_base")

# Encode the text into tokens
tokens = encoding.encode(text)

print(f"Original text: {text}")
print(f"Tokens: {tokens}")
print(f"Number of tokens: {len(tokens)}")

# You can also decode the tokens back to text
decoded_text = encoding.decode(tokens)
print(f"Decoded text: {decoded_text}")
```

To specify a persistent path for `tiktoken`'s cache, you can set the `TIKTOKEN_CACHE_DIR` environment variable. For example, to store the cache in a directory named `my_tiktoken_cache` in your current working directory, you would do the following:

```python
import os

# Define your desired cache directory
custom_cache_dir = "./my_tiktoken_cache"

# Create the directory if it doesn't exist
os.makedirs(custom_cache_dir, exist_ok=True)

# Set the environment variable
os.environ["TIKTOKEN_CACHE_DIR"] = custom_cache_dir

print(f"TIKTOKEN_CACHE_DIR set to: {os.environ.get('TIKTOKEN_CACHE_DIR')}")
print("Future tiktoken model downloads will be cached in this directory.")

# Now, when you call tiktoken.get_encoding(), it will use this path
# For demonstration, let's try getting an encoding again:
import tiktoken
encoding_with_custom_cache = tiktoken.get_encoding("cl100k_base")
print(f"Encoding 'cl100k_base' loaded, using cache at {custom_cache_dir}")
```

Keep in mind that you need to set this environment variable **before** `tiktoken` attempts to download or load any encoding data. If `tiktoken` has already downloaded models to its default location in the current session, they might still be used from there unless you clear the default cache or restart your environment after setting the variable.
